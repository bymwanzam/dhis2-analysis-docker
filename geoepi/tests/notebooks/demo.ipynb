# Demo notebook

```python
data_dir="/x"
out_dir="/y"
```

```python
import json, os
rows=open(os.path.join(data_dir,"pts.csv")).read().split()[1:]
pts=[tuple(map(float,r.split(","))) for r in rows]
os.makedirs(out_dir,exist_ok=True)
json.dump({"type":"FeatureCollection","features":[{"type":"Feature","properties":{"n":v},"geometry":{"type":"Point","coordinates":[x,y]}} for x,y,v in pts]},open(out_dir+"/p.geojson","w"))
json.dump({"points":[{"lat":y,"lng":x,"value":v} for x,y,v in pts]},open(out_dir+"/h.heat.json","w"))
json.dump({"layers":[{"id":"p","title":"Pts","kind":"points","file":"p.geojson"},{"id":"h","title":"Heat","kind":"heat","file":"h.heat.json"}]},open(out_dir+"/layers.json","w"))
print("ok",len(pts))
os.makedirs(out_dir+"/handoff",exist_ok=True); open(out_dir+"/handoff/panel.rds","w").write("x")
```
